# RootIQ - 01: Exploratory Data Analysis
This notebook inspects the raw operational telemetry: logs, metrics, traces, and service topology.

```python
import pandas as pd
import matplotlib.pyplot as plt

df_metrics = pd.read_csv('../data/raw/metrics/telemetry_metrics.csv')
df_logs = pd.read_csv('../data/raw/logs/telemetry_logs.csv')
df_traces = pd.read_csv('../data/raw/traces/telemetry_traces.csv')

print('Metrics shape:', df_metrics.shape)
print('Logs shape:', df_logs.shape)
print('Traces shape:', df_traces.shape)
df_metrics.head()
```

### Statistical Summary of Operational Metrics

```python
df_metrics.groupby('service')[['cpu_usage', 'memory_usage', 'latency_ms', 'error_rate']].describe()
```

### Distribution of Latency Across Services

```python
plt.figure(figsize=(10, 4))
for svc, grp in df_metrics.groupby('service'):
    plt.plot(pd.to_datetime(grp['timestamp']), grp['latency_ms'], label=svc)
plt.xlabel('Timestamp')
plt.ylabel('Latency (ms)')
plt.title('Telemetry Latency Spikes Over Time')
plt.legend()
plt.show()
```
